```python
import re

def parse_selfeval(value):
    if pd.isna(value):
        return "neutral"

    numbers = re.findall(r"\d", str(value))
    if not numbers:
        return "neutral"

    avg = round(sum(map(int, numbers)) / len(numbers))

    EMPATHETIC_SELF_EVAL = {
        1: "sadness",
        2: "anger",
        3: "fear",
        4: "joy",
        5: "neutral"
    }

    return EMPATHETIC_SELF_EVAL.get(avg, "neutral")
```

```python
import pandas as pd
import os
import re

# =============================
# PATHS
# =============================
BASE_DIR = "."

DATASETS = {
    "DailyDialog": [
        "DailyDialog/train.csv",
        "DailyDialog/validation.csv",
        "DailyDialog/test.csv"
    ],
    "EmpatheticDialogues": [
        "EmpatheticDialogues/train.csv",
        "EmpatheticDialogues/valid.csv",
        "EmpatheticDialogues/test.csv"
    ],
    "MELD": [
        "MELD/train_sent_emo.csv",
        "MELD/test_sent_emo.csv"
    ]
}

# =============================
# PLUTCHIK KEYWORD SIGNALS
# =============================
PLUTCHIK_KEYWORDS = {
    "Joy": [
        "happy", "glad", "excited", "love", "great", "awesome", "relieved",
        "pleased", "fun", "enjoy"
    ],
    "Sadness": [
        "sad", "down", "cry", "lost", "hurt", "depressed", "upset",
        "heartbroken", "miss"
    ],
    "Anger": [
        "angry", "mad", "furious", "annoyed", "hate", "irritated",
        "frustrated", "rage"
    ],
    "Fear": [
        "scared", "afraid", "fear", "worried", "nervous", "terrified",
        "anxious", "panic"
    ],
    "Surprise": [
        "surprised", "shocked", "unexpected", "wow", "suddenly",
        "amazed"
    ],
    "Disgust": [
        "disgusting", "gross", "nasty", "repulsive", "sick",
        "creepy"
    ]
}

# =============================
# DATASET LABEL MAPPING (SOFT)
# =============================
DATASET_EMO_MAP = {
    "joy": "Joy",
    "happy": "Joy",
    "sadness": "Sadness",
    "sad": "Sadness",
    "anger": "Anger",
    "angry": "Anger",
    "fear": "Fear",
    "surprise": "Surprise",
    "disgust": "Disgust"
}

EMPATHETIC_SELF_EVAL = {
    1: "Sadness",
    2: "Anger",
    3: "Fear",
    4: "Joy",
    5: "Neutral"
}

# =============================
# CLEAN TEXT
# =============================
def clean_text(t):
    if not isinstance(t, str):
        return None
    t = t.lower().strip()
    t = re.sub(r"\s+", " ", t)
    return t if len(t.split()) >= 3 else None

# =============================
# PLUTCHIK DETECTOR
# =============================
def detect_plutchik(text):
    scores = {}
    for emo, words in PLUTCHIK_KEYWORDS.items():
        scores[emo] = sum(1 for w in words if w in text)

    best_emo = max(scores, key=scores.get)
    if scores[best_emo] == 0:
        return None  # no confident emotion

    return best_emo

# =============================
# COLLECT DATA
# =============================
rows = []

# -------- DailyDialog --------
for path in DATASETS["DailyDialog"]:
    if not os.path.exists(path):
        continue

    df = pd.read_csv(path)
    for _, r in df.iterrows():
        text = clean_text(r["dialog"])
        if text is None:
            continue

        label = str(r["emotion"]).lower()
        base_emo = DATASET_EMO_MAP.get(label)

        detected = detect_plutchik(text)
        final_emo = detected if detected else base_emo

        if final_emo is None or final_emo == "Neutral":
            continue

        rows.append({
            "text": text,
            "plutchik_emotion": final_emo,
            "source_dataset": "DailyDialog"
        })

# -------- EmpatheticDialogues --------
for path in DATASETS["EmpatheticDialogues"]:
    if not os.path.exists(path):
        continue

    df = pd.read_csv(
        path,
        engine="python",
        on_bad_lines="skip",
        dtype=str
    )

    for _, r in df.iterrows():
        text = clean_text(r.get("utterance"))
        if text is None:
            continue

        raw = r.get("selfeval")
        try:
            score = int(raw.split("|")[0].split("_")[0])
        except:
            continue

        base_emo = EMPATHETIC_SELF_EVAL.get(score)
        detected = detect_plutchik(text)

        final_emo = detected if detected else base_emo
        if final_emo is None or final_emo == "Neutral":
            continue

        rows.append({
            "text": text,
            "plutchik_emotion": final_emo,
            "source_dataset": "EmpatheticDialogues"
        })

# -------- MELD --------
for path in DATASETS["MELD"]:
    if not os.path.exists(path):
        continue

    df = pd.read_csv(
        path,
        engine="python",
        on_bad_lines="skip"
    )

    for _, r in df.iterrows():
        text = clean_text(r.get("Utterance"))
        if text is None:
            continue

        label = str(r.get("Emotion")).lower()
        base_emo = DATASET_EMO_MAP.get(label)
        detected = detect_plutchik(text)

        final_emo = detected if detected else base_emo
        if final_emo is None or final_emo == "Neutral":
            continue

        rows.append({
            "text": text,
            "plutchik_emotion": final_emo,
            "source_dataset": "MELD"
        })

# =============================
# FINAL DATAFRAME
# =============================
df = pd.DataFrame(rows)

print("\nEmotion distribution BEFORE balance:")
print(df["plutchik_emotion"].value_counts())

# =============================
# BALANCE (NO FAKE NEUTRAL)
# =============================
TARGET = 12000
balanced = []

for emo in df.plutchik_emotion.unique():
    part = df[df.plutchik_emotion == emo]
    if len(part) >= TARGET:
        balanced.append(part.sample(TARGET, random_state=42))
    else:
        balanced.append(part.sample(TARGET, replace=True, random_state=42))

df_final = pd.concat(balanced).sample(frac=1).reset_index(drop=True)

df_final.to_csv("plutchik_high_quality_dataset.csv", index=False)

print("\nFINAL DISTRIBUTION:")
print(df_final["plutchik_emotion"].value_counts())
print("\nSaved: plutchik_high_quality_dataset.csv")
```

```python
import pandas as pd
import os
import re
from collections import defaultdict

# =========================
# DATASET FILE PATHS
# =========================
DATA_PATHS = [
    ("DailyDialog", "DailyDialog/train.csv"),
    ("DailyDialog", "DailyDialog/validation.csv"),
    ("DailyDialog", "DailyDialog/test.csv"),
    ("EmpatheticDialogues", "EmpatheticDialogues/train.csv"),
    ("EmpatheticDialogues", "EmpatheticDialogues/valid.csv"),
    ("EmpatheticDialogues", "EmpatheticDialogues/test.csv"),
    ("MELD", "MELD/train_sent_emo.csv"),
    ("MELD", "MELD/test_sent_emo.csv"),
]

# =========================
# PLUTCHIK SEMANTIC KEYWORDS
# =========================
PLUTCHIK_SEMANTIC = {
    "Serenity": ["calm", "peaceful", "content", "relaxed"],
    "Joy": ["happy", "glad", "pleased", "smile", "cheerful", "delighted"],
    "Ecstasy": ["thrilled", "excited", "overjoyed", "amazing", "fantastic"],

    "Pensiveness": ["thoughtful", "quiet", "reflecting"],
    "Sadness": ["sad", "down", "cry", "upset", "gloomy"],
    "Grief": ["heartbroken", "devastated", "loss", "mourning"],

    "Annoyance": ["annoyed", "irritated", "frustrated"],
    "Anger": ["angry", "mad", "furious"],
    "Rage": ["rage", "hate", "explode"],

    "Apprehension": ["uneasy", "concerned", "nervous"],
    "Fear": ["scared", "afraid", "fear"],
    "Terror": ["terrified", "panic", "horror"],

    "Distraction": ["confused", "huh", "distracted"],
    "Surprise": ["surprised", "unexpected", "shocked"],
    "Amazement": ["wow", "astonished", "amazed"],

    "Boredom": ["bored", "meh", "uninterested"],
    "Disgust": ["gross", "disgusting", "yuck"],
    "Loathing": ["repulsive", "sickening", "detest"],

    "Acceptance": ["okay", "fine", "alright"],
    "Trust": ["trust", "rely", "safe"],
    "Admiration": ["respect", "inspired", "admire"],

    "Interest": ["interested", "curious", "intrigued"],
    "Anticipation": ["hope", "expect", "looking forward"],
    "Vigilance": ["alert", "watchful", "careful"]
}

INTENSIFIERS = ["very", "extremely", "so", "too", "really", "!!!"]

# =========================
# MERGE MAP TO PLUTCHIK PRIMARY
# =========================
MERGE_MAP = {
    "Serenity": "Joy",
    "Ecstasy": "Joy",
    "Pensiveness": "Sadness",
    "Grief": "Sadness",
    "Annoyance": "Anger",
    "Rage": "Anger",
    "Apprehension": "Fear",
    "Terror": "Fear",
    "Distraction": "Surprise",
    "Amazement": "Surprise",
    "Loathing": "Disgust",
    "Boredom": "Disgust",
    "Acceptance": "Trust",
    "Admiration": "Trust",
    "Interest": "Anticipation",
    "Vigilance": "Anticipation"
}

# =========================
# DATASET LABEL MAPS
# =========================
DATASET_EMO_MAP = {
    "joy": "Joy",
    "happy": "Joy",
    "sadness": "Sadness",
    "sad": "Sadness",
    "anger": "Anger",
    "angry": "Anger",
    "fear": "Fear",
    "surprise": "Surprise",
    "disgust": "Disgust",
}

EMPATHETIC_SELF_EVAL = {
    1: "Sadness",
    2: "Anger",
    3: "Fear",
    4: "Joy",
    5: "Neutral"
}

# =========================
# CLEAN TEXT FUNCTION
# =========================
def clean_text(text):
    if not isinstance(text, str):
        return None
    text = text.lower().strip()
    text = re.sub(r"\s+", " ", text)
    if len(text.split()) < 3:
        return None
    return text

# =========================
# MULTI-EMOTION SEMANTIC MATCH
# =========================
def multi_semantic_match(text):
    scores = defaultdict(int)
    for emo, keywords in PLUTCHIK_SEMANTIC.items():
        for kw in keywords:
            if kw in text:
                scores[emo] += 2
    for intens in INTENSIFIERS:
        if intens in text:
            for emo in scores:
                scores[emo] += 1
    # Keep only emotions with score > 0
    return {k:v for k,v in scores.items() if v>0}

# =========================
# LOAD AND PROCESS DATASETS
# =========================
rows = []

for dataset, path in DATA_PATHS:
    if not os.path.exists(path):
        print("Missing:", path)
        continue

    df = pd.read_csv(path, engine="python", on_bad_lines="skip", dtype=str)

    for _, r in df.iterrows():
        text = None
        raw_labels = []

        # Extract text and raw labels
        if "dialog" in r:
            text = clean_text(r["dialog"])
            lbl = str(r.get("emotion", "")).lower()
            if lbl in DATASET_EMO_MAP:
                raw_labels.append(DATASET_EMO_MAP[lbl])
        elif "utterance" in r:
            text = clean_text(r["utterance"])
            lbl = r.get("selfeval")
            try:
                lbl = int(str(lbl).split("|")[0].split("_")[0])
                lbl = EMPATHETIC_SELF_EVAL.get(lbl)
                if lbl and lbl.lower() != "neutral":
                    raw_labels.append(lbl)
            except:
                pass
        elif "Utterance" in r:
            text = clean_text(r["Utterance"])
            lbl = str(r.get("Emotion", "")).lower()
            if lbl in DATASET_EMO_MAP:
                raw_labels.append(DATASET_EMO_MAP[lbl])

        if not text:
            continue

        # Multi semantic match
        sem_matches = multi_semantic_match(text)
        sem_emotions = list(sem_matches.keys())

        # Combine dataset labels + semantic matches
        all_emotions = list(set(raw_labels + sem_emotions))
        if not all_emotions:
            continue

        # Merge to primary Plutchik emotions
        final_emotions = []
        for emo in all_emotions:
            if re.match(r"[\[\]\d]", str(emo)):
                continue
            final_emotions.append(MERGE_MAP.get(emo, emo))

        if not final_emotions:
            continue

        rows.append({
            "text": text,
            "plutchik_emotion": final_emotions,   # list of emotions
            "source_dataset": dataset
        })

# =========================
# FINAL DATAFRAME
# =========================
df_final = pd.DataFrame(rows)

# Convert list to string for CSV
df_final["plutchik_emotion"] = df_final["plutchik_emotion"].apply(lambda x: ",".join(x))

print("\nFINAL EMOTION DISTRIBUTION (multi-emotion):")
print(df_final["plutchik_emotion"].value_counts())

df_final.to_csv("plutchik_primary_multi_emotion_dataset.csv", index=False)
print("\nSaved: plutchik_primary_multi_emotion_dataset.csv")
```

```python
import torch
import torch.nn as nn
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    BartTokenizer,
    BartForConditionalGeneration
)
from sentence_transformers import SentenceTransformer

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ======================================
# 1. LOAD MODELS
# ======================================

emotion_tokenizer = AutoTokenizer.from_pretrained(
    "j-hartmann/emotion-english-distilroberta-base"
)
emotion_model = AutoModelForSequenceClassification.from_pretrained(
    "j-hartmann/emotion-english-distilroberta-base"
).to(DEVICE)
emotion_model.eval()

sbert = SentenceTransformer("all-MiniLM-L6-v2", device=DEVICE)

bart_tokenizer = BartTokenizer.from_pretrained("facebook/bart-large")
bart_model = BartForConditionalGeneration.from_pretrained(
    "facebook/bart-large"
).to(DEVICE)
bart_model.eval()

# ======================================
# 2. GRU CONTEXT (NO TRAINING)
# ======================================

class GRUContext(nn.Module):
    def __init__(self, input_dim=384, hidden_dim=256):
        super().__init__()
        self.gru = nn.GRU(input_dim, hidden_dim, batch_first=True)

    def forward(self, x):
        with torch.no_grad():
            _, h = self.gru(x)
        return h.squeeze(0)

context_gru = GRUContext().to(DEVICE)
context_gru.eval()

# ======================================
# 3. EMOTION DETECTION (MULTI-LABEL)
# ======================================

def detect_emotions(text, threshold=0.3):
    inputs = emotion_tokenizer(
        text, return_tensors="pt", truncation=True
    ).to(DEVICE)

    with torch.no_grad():
        logits = emotion_model(**inputs).logits
        probs = torch.sigmoid(logits).squeeze()

    emotions = {}
    for i, p in enumerate(probs):
        if p.item() >= threshold:
            emotions[emotion_model.config.id2label[i]] = round(p.item(), 3)

    return emotions, probs

# ======================================
# 4. SBERT + CONTEXT
# ======================================

def get_context_embedding(text):
    emb = sbert.encode(text, convert_to_tensor=True)
    emb = emb.unsqueeze(0).unsqueeze(0)
    ctx = context_gru(emb)
    return ctx

# ======================================
# 5. RESPONSE GENERATION
# ======================================

def generate_response(text, emotions):
    emo_prefix = " ".join(f"[{e}]" for e in emotions.keys())
    prompt = f"{emo_prefix} {text}"

    inputs = bart_tokenizer(
        prompt, return_tensors="pt", truncation=True
    ).to(DEVICE)

    with torch.no_grad():
        output = bart_model.generate(
            **inputs,
            max_length=70,
            num_beams=4,
            early_stopping=True
        )

    return bart_tokenizer.decode(
        output[0], skip_special_tokens=True
    )

# ======================================
# 6. EMOTION ALIGNMENT CHECK (BEERT STYLE)
# ======================================

def emotion_alignment(user_probs, response):
    _, resp_probs = detect_emotions(response)
    return torch.cosine_similarity(user_probs, resp_probs, dim=0).item()

# ======================================
# 7. FULL PIPELINE
# ======================================

def empathetic_chat(text):
    emotions, user_probs = detect_emotions(text)
    _ = get_context_embedding(text)  # context memory
    response = generate_response(text, emotions)
    score = emotion_alignment(user_probs, response)

    if score < 0.45:
        response = generate_response(
            "Respond empathetically: " + text,
            emotions
        )

    return response, emotions, round(score, 3)

# ======================================
# 8. RUN CHATBOT
# ======================================

if __name__ == "__main__":
    print("\nEmpathetic Chatbot Ready")
    print("Type 'exit' to stop\n")

    while True:
        user_text = input("You: ")
        if user_text.lower() == "exit":
            break

        response, emotions, score = empathetic_chat(user_text)

        print("Detected emotions:", emotions)
        print("Emotion match score:", score)
        print("Bot:", response)
        print()
```

```python
import os
os.environ["HF_HUB_DISABLE_SYMLINKS_WARNING"] = "1"

import torch
from transformers import BartTokenizer, BartForConditionalGeneration
from sentence_transformers import SentenceTransformer
from scipy.spatial.distance import cosine

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# =========================
# LOAD MODELS
# =========================
bart_tokenizer = BartTokenizer.from_pretrained("facebook/bart-large-cnn")
bart_model = BartForConditionalGeneration.from_pretrained(
    "facebook/bart-large-cnn"
).to(DEVICE)

sbert = SentenceTransformer("all-MiniLM-L6-v2")

# =========================
# PLUTCHIK EMOTIONS
# =========================
EMOTION_ANCHORS = {
    "joy": "happy joyful excited content",
    "sadness": "sad lonely depressed hurt",
    "anger": "angry frustrated annoyed mad",
    "fear": "scared afraid anxious worried",
    "surprise": "shocked amazed startled",
    "trust": "safe supported secure",
    "neutral": "calm normal okay"
}

ANCHOR_EMB = {
    e: sbert.encode(t)
    for e, t in EMOTION_ANCHORS.items()
}

EMPATHY_INSTRUCTION = {
    "sadness": "Comfort the user and reassure them warmly.",
    "joy": "Share their happiness and encourage positivity.",
    "anger": "Calm them and validate their feelings.",
    "fear": "Reassure them and help them feel safe.",
    "surprise": "Respond with understanding.",
    "trust": "Strengthen emotional support.",
    "neutral": "Respond politely and friendly."
}

# =========================
# EMOTION DETECTION
# =========================
def detect_emotion(text):
    emb = sbert.encode(text)
    scores = {
        e: 1 - cosine(emb, a)
        for e, a in ANCHOR_EMB.items()
    }
    dominant = max(scores, key=scores.get)
    return dominant, scores

# =========================
# RESPONSE GENERATION
# =========================
def generate_response(user_text, emotion):
    instruction = EMPATHY_INSTRUCTION[emotion]

    prompt = (
        f"Instruction: {instruction}\n"
        f"User says: {user_text}\n"
        f"Empathetic response:"
    )

    inputs = bart_tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=256
    ).to(DEVICE)

    with torch.no_grad():
        output = bart_model.generate(
            **inputs,
            max_new_tokens=60,
            num_beams=5,
            do_sample=True,
            top_p=0.9,
            temperature=0.8,
            repetition_penalty=1.3
        )

    decoded = bart_tokenizer.decode(
        output[0],
        skip_special_tokens=True
    )

    # 🔑 REMOVE PROMPT PART
    return decoded.split("Empathetic response:")[-1].strip()

# =========================
# CHAT LOOP
# =========================
print("\nEmpathetic Chatbot Ready")
print("Type 'exit' to stop\n")

while True:
    user_input = input("You: ").strip()
    if user_input.lower() == "exit":
        break

    emotion, scores = detect_emotion(user_input)
    response = generate_response(user_input, emotion)

    print("\nDetected emotions:", {k: round(v, 3) for k, v in scores.items()})
    print("Dominant emotion:", emotion)
    print("Bot:", response)
    print("-" * 60)
```

```python
import os
os.environ["HF_HUB_DISABLE_SYMLINKS_WARNING"] = "1"

import torch
from transformers import BartTokenizer, BartForConditionalGeneration
from sentence_transformers import SentenceTransformer
from scipy.spatial.distance import cosine

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# =========================
# LOAD MODELS
# =========================
bart_tokenizer = BartTokenizer.from_pretrained("facebook/bart-large-cnn")
bart_model = BartForConditionalGeneration.from_pretrained(
    "facebook/bart-large-cnn"
).to(DEVICE)

sbert = SentenceTransformer("all-MiniLM-L6-v2")

# =========================
# EMOTION SPACE
# =========================
EMOTION_ANCHORS = {
    "joy": "happy joyful excited content",
    "sadness": "sad lonely depressed hurt",
    "anger": "angry frustrated annoyed mad",
    "fear": "scared afraid anxious worried",
    "surprise": "shocked amazed startled",
    "trust": "safe supported secure",
    "neutral": "calm normal okay"
}

ANCHOR_EMB = {k: sbert.encode(v) for k, v in EMOTION_ANCHORS.items()}

EMPATHY_STYLE = {
    "sadness": "comforting and reassuring",
    "joy": "warm and cheerful",
    "anger": "calm and understanding",
    "fear": "gentle and reassuring",
    "surprise": "kind and understanding",
    "trust": "supportive and warm",
    "neutral": "friendly and polite"
}

# =========================
# EMOTION DETECTION
# =========================
def detect_emotion(text):
    emb = sbert.encode(text)
    scores = {
        e: 1 - cosine(emb, a)
        for e, a in ANCHOR_EMB.items()
    }
    return max(scores, key=scores.get), scores

# =========================
# RESPONSE GENERATION (FIXED)
# =========================
def generate_response(user_text, emotion):
    style = EMPATHY_STYLE[emotion]

    prompt = (
        f"Write ONE short empathetic reply.\n"
        f"Tone: {style}.\n"
        f"User message: {user_text}\n"
        f"Reply:"
    )

    inputs = bart_tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=256
    ).to(DEVICE)

    with torch.no_grad():
        output = bart_model.generate(
            **inputs,
            max_new_tokens=30,              # 🔑 VERY IMPORTANT
            num_beams=1,                    # disables dialogue expansion
            do_sample=False,                # deterministic
            repetition_penalty=2.0,         # kills looping
            no_repeat_ngram_size=3,
            early_stopping=True
        )

    decoded = bart_tokenizer.decode(
        output[0],
        skip_special_tokens=True
    )

    # 🔑 HARD CUT AFTER REPLY
    reply = decoded.split("Reply:")[-1].strip()
    reply = reply.split("User")[0].strip()
    reply = reply.split("user")[0].strip()

    return reply

# =========================
# CHAT LOOP
# =========================
print("\nEmpathetic Chatbot Ready")
print("Type 'exit' to stop\n")

while True:
    user_input = input("You: ").strip()
    if user_input.lower() == "exit":
        break

    emotion, scores = detect_emotion(user_input)
    response = generate_response(user_input, emotion)

    print("\nDetected emotions:", {k: round(v, 3) for k, v in scores.items()})
    print("Dominant emotion:", emotion)
    print("Bot:", response)
    print("-" * 60)
```

```python
from dotenv import load_dotenv
import os
from groq import Groq

# Load environment variables from .env
load_dotenv()

# Get Groq API key
GROQ_API_KEY = os.getenv("GROQ_API_KEY")
```

```python
# ===============================
# 1️⃣ IMPORTS
# ===============================
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from groq import Groq

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# ===============================
# 2️⃣ EMOTIONS
# ===============================
EMOTIONS = [
    "joy", "sadness", "anger", "fear",
    "surprise", "trust", "disgust", "anticipation"
]

EMPATHY_PROMPT = {
    "joy": "Respond in a warm and positive way that matches happiness.",
    "sadness": "Respond with emotional support, comfort, and reassurance.",
    "anger": "Respond calmly and help the user feel understood.",
    "fear": "Respond with reassurance and reduce anxiety.",
    "surprise": "Respond with curiosity and gentle engagement.",
    "trust": "Respond in a friendly and supportive way.",
    "disgust": "Respond respectfully and help shift focus calmly.",
    "anticipation": "Respond with encouragement and optimism.",
    "neutral": "Respond in a polite and friendly manner."
}

# ===============================
# 3️⃣ LOAD YOUR EMOTION MODEL
# ===============================
# Recreate the model architecture
emotion_model = AutoModelForSequenceClassification.from_pretrained(
    "bert-base-uncased", num_labels=len(EMOTIONS)
)

# Load your trained state dict
state_dict = torch.load("emotion_model.pt", map_location=DEVICE)

# 🔹 Fix keys to match Hugging Face naming
new_state_dict = {}
for k, v in state_dict.items():
    new_key = k
    if k.startswith("encoder."):
        new_key = "bert." + k[len("encoder."):]  # rename encoder -> bert
    if k.startswith("fc."):
        new_key = "classifier." + k[len("fc."):]  # rename fc -> classifier
    new_state_dict[new_key] = v

# Load the fixed state dict
emotion_model.load_state_dict(new_state_dict)

# Move to device and set eval
emotion_model.to(DEVICE)
emotion_model.eval()

# Load the tokenizer you used for training
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# ===============================
# 4️⃣ GROQ CLIENT SETUP
# ===============================
client = Groq(api_key=GROQ_API_KEY)  # put your Groq API key here

SYSTEM_PROMPT = """
You are an emotional and empathetic chatbot.
Always respond kindly.
Focus only on the user's current message.
Do not rely on past messages.
Acknowledge feelings before giving any response.
Keep responses short and human.
"""

# ===============================
# 5️⃣ EMOTION DETECTION FUNCTION
# ===============================
def detect_emotion(text):
    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True).to(DEVICE)
    
    with torch.no_grad():
        logits = emotion_model(**inputs).logits
        probs = torch.sigmoid(logits).cpu().numpy()[0]
    
    emotion_scores = dict(zip(EMOTIONS, probs))
    # Add neutral as fallback
    emotion_scores["neutral"] = 1 - max(probs)
    
    return emotion_scores

# ===============================
# 6️⃣ CHAT FUNCTION USING GROQ
# ===============================
def chat(user_message):
    emotions = detect_emotion(user_message)
    dominant_emotion = max(emotions, key=emotions.get)
    empathy_instruction = EMPATHY_PROMPT[dominant_emotion]
    
    messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "system", "content": f"Detected emotion: {dominant_emotion}. {empathy_instruction}"},
        {"role": "user", "content": user_message}
    ]
    
    response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=messages,
        temperature=0.8
    )
    
    return response.choices[0].message.content, emotions, dominant_emotion

# ===============================
# 7️⃣ INTERACTIVE LOOP
# ===============================
print("Empathetic Chatbot Ready! Type 'exit' to quit.\n")

while True:
    user_input = input("You: ").strip()
    if user_input.lower() in ["exit", "quit"]:
        print("Bot: Take care. I'm here whenever you need.")
        break
    
    bot_response, emotions, dominant = chat(user_input)
    print(f"\nDetected emotions: {emotions}")
    print(f"Dominant emotion: {dominant}")
    print(f"Bot: {bot_response}\n")
```
